# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

When presenting the number of visitors to a museum over two weeks, using separate plots for each week allows the viewer to clearly see the pattern within each week without overlap, making individual trends easy to identify. However, this approach makes it harder to directly compare the two weeks, since the viewer must mentally align the plots to see changes from one week to the next. On the other hand, a combined plot shows both weeks on the same axes, making week-to-week comparisons immediate and highlighting differences in trends or peaks across days. The drawback of this method is that overlapping lines can create visual clutter, and if the scales of the two weeks differ significantly, one line might dominate the plot and obscure the other. In general, separate plots emphasize within-week patterns, while combined plots emphasize comparisons between weeks.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Simulated sales growth data over 5 months
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales_product_a = [50, 55, 60, 65, 80]
sales_product_b = [40, 42, 45, 50, 55]

df_sales = pd.DataFrame({
    'Month': months,
    'Product A': sales_product_a,
    'Product B': sales_product_b
})

# Plot
fig, ax = plt.subplots(figsize=(8,5))

bar_width = 0.35
x = np.arange(len(months))

# Draw bars
ax.bar(x - bar_width/2, df_sales['Product A'], width=bar_width, color='steelblue', label='Product A')
ax.bar(x + bar_width/2, df_sales['Product B'], width=bar_width, color='lightgray', label='Product B')

# Highlight May for Product A (annotation to the side)
ax.annotate('Strong growth!',
            xy=(x[-1] - bar_width/2, df_sales['Product A'].iloc[-1]),
            xytext=(x[-1] + 0.5, df_sales['Product A'].iloc[-1]),  # move to the side
            arrowprops=dict(facecolor='black', arrowstyle='->'),
            va='center')  # vertical alignment

# Labels and styling
ax.set_xticks(x)
ax.set_xticklabels(months)
ax.set_ylabel("Sales")
ax.set_title("Monthly Sales Comparison (Highlighting Product A)", pad=15)
ax.legend()
plt.show()
```
